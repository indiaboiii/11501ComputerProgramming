問題：撰寫一Python程式，為使用者輸入年份（可能是西元或民國）找尋其相對應的十二生肖。十二生肖是以十二年為一週期，每一年分別以一種動物來表示猴（monkey）、雞（rooster）、狗(dog)、豬(pig)、鼠(rat)、牛(ox)、虎(tiger)、兔(rabbit)、龍(dragon)、蛇（snake）、馬(horse)，以及羊(sheep) 。

````python
def get_zodiac():
    # 十二生肖列表，以鼠年作為對應基準索引 0 (西元 2020 年 / 民國 109 年為鼠年)
    zodiacs = ['鼠 (rat)', '牛 (ox)', '虎 (tiger)', '兔 (rabbit)',
               '龍 (dragon)', '蛇 (snake)', '馬 (horse)', '羊 (sheep)',
               '猴 (monkey)', '雞 (rooster)', '狗 (dog)', '豬 (pig)']

    user_input = input("請輸入年份（可輸入西元或民國，例如 2026 或 115）：").strip()

    try:
        year = int(user_input)
    except ValueError:
        print("輸入格式錯誤，請輸入有效的整數年份。")
        return

    # 判斷是否為民國年份（若小於 1000 通常視為民國，或可根據實際需求調整判斷邏輯）
    # 這裡以小於 1911 年的數值，或若使用者明確知道小於 1000 視為民國
    # 習慣上民國年份多小於 200，西元年大於 1900
    if year < 1000:
        se_year = year + 1911
        print(f"您輸入的是民國 {year} 年（相當於西元 {se_year} 年）")
    else:
        se_year = year
        print(f"您輸入的是西元 {se_year} 年")

    # 以西元 2020 年（鼠年）為基準計算餘數
    index = (se_year - 2020) % 12

    print(f"對應的十二生肖為：{zodiacs[index]}")

if __name__ == "__main__":
    get_zodiac()
````

   ## 0.1 的二進位表示法與計算過程

   ### 將十進位純小數轉換為二進位，採用「乘 2 取整數」法：

   - $0.1 \times 2 = 0.2$ （整數部分：$0$）
   
   - $0.2 \times 2 = 0.4$ （整數部分：$0$）
   
   - $0.4 \times 2 = 0.8$ （整數部分：$0$）
   
   - $0.8 \times 2 = 1.6$ （整數部分：$1$，小數部分：$0.6$）
   
   - $0.6 \times 2 = 1.2$ （整數部分：$1$，小數部分：$0.2$）

   此時小數部分再度回到 $0.2$，代表接下來的位數會進入無限循環（循環序列為 $0011$）


   ## 看到的現象
   #### 因為電腦的記憶體/暫存空間有限（例如 Python 的浮點數採用 64 位元雙精度），無法無限儲存這些位數，系統必須在某個位置進行截斷（捨入）

   #### 因此，當我們在 Python 中執行以下程式碼時：

- Python
   
     print(0.1 + 0.2)

   畫面上不會完美印出 0.3，而是會印出：

   Plaintext


   0.30000000000000004

   這就是因為 $0.1$ 和 $0.2$ 在轉成二進位儲存時就已經產生了微小的捨入誤差，經過加總後這個誤差就被顯現出來了


2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
  
  
   ## 當小數轉換成二進位後變成無限循環小數時，Python（以及多數程式語言）在底層主要透過以下兩個機制來處理有限的記憶體空間：

  - ### IEEE 754 雙精度標準：


     Python 的浮點數（float）在底層採用了 IEEE 754 標準的 64 位元（8 bytes）雙精度格式來儲存數字。它將 64 個位元固定分配給符號位、指數位與尾數（有效數字）。當二進位位數超出可記錄的上限時，系統會強制進行截斷與捨入，將無限位數收斂為有限位數


  - ### 四捨六入五成雙（Round half to even）：

     #### 在進行捨入以決定要進位還是捨去時，IEEE 754 標準採用了「四捨六入五成雙」的規則：

     - 大於 5 則進位，小於 5 則捨去
     - 若尾數剛好等於 5：則會檢查 5 前面那一位數是奇數還是偶數。若是偶數則直接捨去（保持偶數），若是奇數則進位（使其變成偶數）。這個設計能有效平衡大量計算時的累積誤差，避免傳統四捨五入造成整體數據持續偏大的問題

  ## Python 如何知道浮點數的精度？

  - ### 底層機制：


     Python 本身是透過底層的 C 語言（通常是 double 型態）以及作業系統/硬體架構來決定浮點數的精度。它會直接向硬體（例如 CPU 的浮點運算單元 FPU）查詢當前系統所支援的 IEEE 754 雙精度浮點數極限，並透過內建的 sys 模組將這些資訊包裝成 sys.float_info 供開發者調用

  ## 執行程式碼與輸出結果說明
當您在 Python 中執行這段程式碼：

```python
import sys
print(sys.float_info)
```

系統會印出一個包含多項參數的結構化物件（Named Tuple），裡面記錄了當前直譯器下浮點數的各項極限與精度數據。其中幾個最核心且常被問到的欄位意義如下：

  - dig（十進位有效精度）：

     含意：代表浮點數能保證精確的十進位有效數字位數。在多數標準 64 位元系統中，這個值通常是 15，代表大約有 15 位十進位數字是絕對準確的
- epsilon（機器精度 / 最小差值）：

     含意：這是 1.0 與大於 1.0 的下一個可表示浮點數之間的最小差值（約為 $2.22 \times 10^{-16}$）。它代表系統能分辨的最小精度極限，小於這個差距的浮點數變化在電腦裡會被視為零

- mant_dig（尾數二進位位數）：

     含意：浮點數尾數（有效數字部分）所佔的二進位位數。在雙精度中通常是 53 位（包含 IEEE 754 隱含的領頭 1）



４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```

## 解決方法一：使用 round() 函數四捨五入

利用 Python 內建的 round() 函數，指定要取到的小數點位數，把多餘的誤差尾數修掉：
```
result = round(0.1 + 0.2, 1)
print(result)  
```

## 解決方法二：使用 decimal 模組（最精確）

Python 內建了專門用來進行高精度十進位運算的 decimal 模組，它能完美避免二進位轉換帶來的誤差：
```
from decimal import Decimal

result = Decimal('0.1') + Decimal('0.2')
print(result)
```

## 解決方法三：使用字串格式化（適合用於畫面上顯示與輸出）
透過字串格式化指定小數點位數，只顯示到想要的位數：
```
result = 0.1 + 0.2
print(f"{result:.1f}")
```

   ## 金融領域（Financial Applications）的分析

- 結論：前面提過的 round() 絕對不是好方法，而應該使用 decimal 模組

 - 原因：

   - 不能有絲毫誤差：金融計算（如利息、稅金、匯率轉換）涉及金錢，哪怕只有 $0.00001$ 的誤差，累積起來也會造成帳目不符或法律責任

   - 四捨五入的陷阱：前面提到的 round() 或者是傳統的四捨五入，在大量交易累積下會產生「統計偏差」（讓數字偏大）。金融業通常要求更嚴謹的捨入規範（例如前面提到的「四捨六入五成雙」或強制指定小數位數的定點數運算）。因此在金融上，使用高精度的 decimal.Decimal 才是業界標準的好方法

   ## 科學計算（Scientific Computing）的分析

- 結論：不管是 round() 或是 decimal，通常都不是最佳解
- 原因：

   - 效能與速度的考量：科學計算（如氣象預測、物理模擬、大數據分析）往往需要處理億萬等級的浮點數運算。decimal 或高精度運算的執行速度遠比硬體原生支援的 float（64-bit 浮點數）慢很多，會導致計算時間過長

   - 誤差的控制與接受度：在科學計算中，物理測量本身就帶有誤差（例如儀器誤差），因此科學家們更關注的是「數值穩定性（Numerical Stability）」與演化過程中的誤差累積，通常會透過設計良好的演算法（例如避免相近數字相減導致的「有效數字失真」）來降低影響，而不是盲目地在每一步都去做四捨五入
