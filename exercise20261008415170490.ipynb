使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出$nn$乘法表。格式如Tranclass的圖片Exercise20261008.png，以$n=5$為例。

1. 只能使用for。
3. 字串格式化只能使用 f-string 對齊。

```python
# 取得使用者輸入的正整數 n (1 <= n <= 9)
n = int(input("請輸入正整數n，其中1 <= n <= 9: "))

# 根據 n 的值計算二進位顯示所需的位數
w1 = n
w2 = 2 * n

# 使用 for 迴圈產生 n×n 乘法表
for i in range(1, n + 1):
    row_str = ""
    for j in range(1, n + 1):
        # 直接使用 f-string 的 :0{w}b 格式化整數為指定寬度的二進位並補零
        expression = f"{i:0{w1}b}×{j:0{w1}b}={i * j:0{w2}b}"

        # 將每個算式組合，欄位之間保持適當間距
        row_str += f"{expression}   "

    # 印出每一行的乘法表結果
    print(row_str.rstrip())
```

使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出二進位$nn$乘法表。格式如如Tranclass的圖片Exercise20261008.png，以$n=5$為例。                                                                                                         
                                                                                                            
1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。                                                                       

```python
# 取得使用者輸入的正整數 n (1 <= n <= 9)
n = int(input("請輸入正整數n，其中1 <= n <= 9: "))

# 根據 n 的值計算二進位顯示所需的位數
w1 = n
w2 = 2 * n

# 使用 while 迴圈替代 for 迴圈來控制列 (i)
i = 1
while i <= n:
    row_str = ""
    j = 1
    # 使用巢狀的 while 迴圈來控制欄 (j)
    while j <= n:
        # 使用 f-string 進行二進位格式化與補零對齊
        expression = f"{i:0{w1}b}×{j:0{w1}b}={i * j:0{w2}b}"

        # 將每個算式組合，欄位之間保持適當間距
        row_str += f"{expression}   "
        j += 1

    # 印出每一行的乘法表結果
    print(row_str.rstrip())
    i += 1
```
