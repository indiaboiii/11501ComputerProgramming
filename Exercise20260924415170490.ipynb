## 第四部分：深入探討各種型別

請針對第三部分中的每一種型別回答：

1. 它是否為 Python 的內建型別？以上皆是

2. 對應的 CPython 物件結構名稱為何？int:PyLongObject, float:PyFloatObject, complex:PyComplexObject, set:PySetObject, bool:PyBoolObject, list:PyListObject

3. 它如何儲存資料？

   int:透過變動長度的 digit 陣列來儲存數值，並包含正負號標記

   float:內部直接使用 C 語言的雙精度浮點數（double）來儲存數值

   complex:內部包含兩個 C 語言的 double 欄位，分別儲存實部（real）與虛部（imag）

   set:基於雜湊表（Hash Table）儲存，內部記錄元素的雜湊值、鍵值指標與狀態
   
   bool:全域僅存在 Py_True 與 Py_False 兩個共用實例（Singleton），本質上是帶有特定數值的整數
   
   list:內部使用一個指向 Python 物件指標陣列（PyObject **ob_item）的指標來儲存資料

4. 是否有固定大小限制？

   int,set,list無大小限制float固定（通常為8位元組）
   
   complex大小固定（由兩個double組成）
   
   bool固定為整數物件的大小

5. 若資料規模持續增加，如何擴充儲存空間？

   int:當數值超出原本的digit陣列容量時，CPython會自動動態配置更多的記憶體空間來容納更多的位數
   
   float:小固定，無法擴充
   
   complex:大小固定，不需擴充
   
   set:當雜湊表達到一定的負載係數時，會自動建立一個更大的新雜湊表，並將舊元素重新雜湊（Re-hash）並搬遷進去
  
   bool:只有固定的兩個全域實例，不涉及空間擴充
   
   list:當串列空間不足時，會預先配置額外的空間（over-allocation），以倍數或特定比例擴增底層陣列容量，減少頻繁呼叫記憶體分配的開銷

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？

   Python不會把整數限制在傳統 32 位元或 64 位元的暫存器大小內，而是根據數字的大小動態配置記憶體來容納更多的位數
2. Python 的 `int` 是否具有固定的位元數限制？否


3. 什麼是 Arbitrary Precision（任意精度）？

   一種資料表示法，允許數值的大小只受限於電腦可用的記憶體總量
4. CPython 使用什麼物件來實作 `int`？

   使用 PyLongObject
5. Python 整數大小真正的限制是什麼？

   真正的限制是系統可用的實體記憶體大小。當數字大到把電腦的記憶體全部耗盡時，就會引發記憶體不足錯誤

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？

   因為電腦底層是使用二進位像，0.1和0.2這類十進位小數，轉換成二進位時是無限循環小數，限於記憶體空間有限，電腦必須在某個地方進行四捨五入，因此運算後會累積微小的誤差
2. 浮點數是否能精確表示所有小數？

   不能。只能精確表示某些能被二進位整除的分數（例如0.5、0.25等），絕大多數的十進位小數都只能用「最接近的近似值」來表示。
3. 什麼是浮點數的精度（precision）？

   精度指的是浮點數能夠區分相鄰數值的細膩程度（即有效數字的位數）。在電腦中，精度主要由儲存尾數的位元數決定，位元數越多，能記錄的小數位數就越精確
4. 什麼是浮點數的範圍（range）？

   範圍指的是浮點數能夠表示的最大值與最小值的跨度（從極接近0的微小數字，到非常巨大的天文數字）。這主要由儲存指數（Exponent）的位元數來決定
5. 精度與範圍有何不同？

   精度關注的是「精細度」：數字與數字之間可以切得有多細、有效數字有多少位

   範圍關注的是「大小極限」：這個型別可以容納多大或多小的極端數值（會不會發生溢位 Overflow 或下溢 Underflow）

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？通常使用浮點數，例如 FP32、FP16 或 BF16 等型別


2. 什麼是 FP64？FP64（64-bit Floating-point，雙精度浮點數）佔用8個位元組（64位元）的記憶體。它具有極高的精確度與極大的範圍，但在AI訓練中較少作為模型預設權重，因為計算速度慢且消耗大量記憶體


3. 什麼是 FP32？FP32（32-bit Floating-point，單精度浮點數）佔用 4 個位元組（32 位元）的記憶體。它是過去深度學習中最常用的標準精度，兼顧了計算精度與效能


4. 什麼是 FP16？FP16（16-bit Floating-point，半精度浮點數）佔用 2 個位元組（16 位元）的記憶體。它能大幅減少記憶體佔用並加速運算，但因為數值範圍較小，訓練時容易發生數值下溢（Underflow）或溢位，通常需要混合精度訓練（Mixed Precision）來克服

5. 什麼是 BF16？BF16（Brain Floating-point）同樣佔用 2 個位元組（16 位元），但它把較多的位元分配給「指數（範圍）」，較少的位元分配給「尾數（精度）」。它的動態範圍與 FP32 相同，因此在現代 AI 模型訓練中非常受歡迎，能避免 FP16 容易溢位、又兼具高效能的優點


6. 為什麼大型AI模型不直接使用Python的`int`作為權重？Python的int（PyLongObject）是任意精度整數，會隨著數字變大而動態增加記憶體開銷，且每個物件本身在 CPython 中都有額外的標頭開銷，非常耗費記憶體

   AI 模型需要依靠 GPU 進行大量的矩陣乘法運算，GPU 是針對硬體層級的固定大小浮點數（如 FP32、FP16、BF16 或 INT8）進行高度平行加速設計的，無法直接、有效率地處理 Python 動態大小的 int 物件

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？

   這代表整數的大小與位數不會被固定在硬體暫存器的限制，中當數值越來越大時，系統會自動動態配置更多記憶體來容納，不會發生數值溢位
2. 它是否代表可以儲存無限大的整數？

   並不是真的無限大，它的極限完全取決於電腦當前可用的實體記憶體（RAM）總量。
3. CPython 如何實現這項特性？

   CPython 在底層透過 PyLongObject 結構來實作整數，它內部使用一個變動長度的陣列（數位陣列）來分段儲存大整數的數值，當計算產生的數字超過原本陣列容量時，直譯器會自動透過動態記憶體分配（擴充電腦記憶體空間）來容納更多位數

回答：

1. 為什麼 `int` 幾乎沒有固定上限？

2. 為什麼 `float` 卻存在範圍限制？

3. 為什麼 `float` 可能產生誤差？

   因為電腦底層是使用二進位來表示數值，許多十進位的小數（例如 0.1）無法在二進位中被精確、有限地表達

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。

   Python 的整數（Arbitrary Precision / 正確性優先）：以「確保計算結果絕對正確」為核心，不允許發生數值溢位或精度丟失，會隨著數值大小動態調整記憶體空間

   AI 模型的 FP16、BF16（Limited Precision / 效率優先）：以「追求極致運算效能與節省記憶體」為核心，在可接受的誤差範圍內使用固定且較小的位元數，以換取高度平行的運算速度
2. 各自的優點與缺點。

   Python 整數（Arbitrary Precision）

   優點：絕對不會發生整數溢位（Overflow），大數運算結果 100% 精確

   缺點：記憶體開銷大、結構複雜，運算速度相對較慢，不適合大規模平行計算

   AI 模型 FP16 / BF16（Limited Precision）

   優點：記憶體佔用極小（僅 2 bytes），能大幅降低硬體頻寬壓力，並利用 GPU 加速實現極高的運算吞吐量

   缺點：存在捨入誤差，且因為動態範圍或精度較小，訓練時較容易面臨數值下溢（Underflow）或溢位問題
3. 為什麼大型語言模型通常選擇較低精度的表示方式？

   大型語言模型（LLM）動輒擁有數百億甚至數千億個參數與權重。若全部使用高精度格式，模型檔案會變得極為巨大，一般伺服器或 GPU 的記憶體根本無法載入
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？

   FP64 佔用 64 位元（8 bytes），是 FP16/BF16（16 位元 / 2 bytes）的四倍，原本能載入模型的硬體將因記憶體不足而無法運行，且硬體在處理雙精度浮點數（FP64）時的平行計算效能遠低於低精度浮點數，這會導致模型訓練與推論速度變得極其緩慢，消耗龐大的時間與電力成本
